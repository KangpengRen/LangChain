# `invode`调用的使用

## 1. `invoke`的使用

### 1.1. 文本输入

```python
import os

from dotenv import load_dotenv
from langchain_deepseek import ChatDeepSeek

# 读取.env配置文件中的信息，相关环境变量以.env文件中优先
load_dotenv(override=True)
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 模型初始化
llm_deepseek = ChatDeepSeek(
    model="deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    api_base=DEEPSEEK_BASE_URL
)
```

```python
response = llm_deepseek.invoke("翻译如下汉字：你好，世界！")
print(response)
```

### 1.2. 字典列表

```python
messages = [
    {"role": "system", "content": "你是一个专业的数学老师"},
    {"role": "user", "content": "帮我解释一下什么是斐波拉契数列"}
]

response = llm_deepseek.invoke(messages)
print(response)
```

```python
messages = [
    {"role": "system", "content": "你是一个专业的数学老师"},
    {"role": "user", "content": "1 + 1 = ?"},
    {"role": "assistant", "content": "3"},
    {"role": "user", "content": "我刚才问了什么问题？"},
]

response = llm_deepseek.invoke(messages)
print(response)
```

如果不传递历史，AI会“失忆”

```python
messages1 = [
    {"role": "system", "content": "你是一个非常友好的AI助手"},
    {"role": "user", "content": "你好，我叫小明"}
]

# 第一次对话
response1 = llm_deepseek.invoke(messages1)
print(response1)

messages2 = [
    {"role": "user", "content": "我叫什么名字？"}
]

# 第二次对话
response2 = llm_deepseek.invoke(messages2)
print(response2)
```

作为对比，传递记忆

```python
conversation = [
    {"role": "system", "content": "你是一个非常友好的AI助手"},
    {"role": "user", "content": "你好，我叫小明"}
]

response1 = llm_deepseek.invoke(conversation)
print(response1)

# 添加记忆
conversation.append({"role": "assistant", "content": response1.content})
conversation.append({"role": "user", "content": "我叫什么名字？"})

response2 = llm_deepseek.invoke(conversation)
print(response2)
```

### 1.3. 消息列表

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(content="你是一个专业的数学老师"),
    HumanMessage(content="帮我解释一下什么是斐波拉契数列")
]

response = llm_deepseek.invoke(messages)
print(response)
```

传递记忆

```python
from langchain_core.messages import AIMessage

messages = [
    SystemMessage(content="你是一个专业的数学老师"),
    HumanMessage(content="帮我解释一下什么是斐波拉契数列")
]

response = llm_deepseek.invoke(messages)
print(response)

messages.append(AIMessage(content=response1.content))
messages.append(HumanMessage(content="我刚才问的什么问题"))

response = llm_deepseek.invoke(messages)
print(response)
```

## 2. `invoke`返回值

```python
response = llm_deepseek.invoke("用一句话解释什么是AI")

# 1. 直接获取回答内容
print("AI的回复为：", response.content)

# 2. 获取响应元数据
metadata = response.response_metadata
print(f"使用的模型是：{metadata['model_name']}")
print(f"结束的原因是：{metadata['finish_reason']}")
print(f"模型提供商是：{metadata['model_provider']}")

# 3. 获取Token使用情况
usage = metadata.get("token_usage", {})
print(f"输入tokens：{usage.get("prompt_tokens")}")
print(f"输出tokens：{usage.get("completion_tokens")}")
print(f"总计tokens：{usage.get("total_tokens")}")

# 4. 消息ID
print(f"消息ID：{response.id}")
```
