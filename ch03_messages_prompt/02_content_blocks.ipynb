# `content`和`blocks`的使用

保存多模态

```python
import base64
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

# 读取.env配置文件中的信息，相关环境变量以.env文件中优先
load_dotenv(override=True)
CLOSEAI_API_KEY = os.getenv("CLOSEAI_API_KEY")
CLOSEAI_BASE_URL = os.getenv("CLOSEAI_BASE_URL")

model = init_chat_model(
    model="gpt-5.4-mini",
    model_provider="openai",
    temperature=2,
    api_key=CLOSEAI_API_KEY,
    base_url=CLOSEAI_BASE_URL
)


def encode_image(img_path, img_type="jpeg"):
    with open(img_path, "rb") as img_file:
        return f"data:image/{img_type};base64,{base64.b64encode(img_file.read()).decode("utf-8")}"


# 图像路径
img_path = "avatar.jpg"

base64_image = encode_image(img_path)

response = model.invoke(
    [
        HumanMessage(
            content=[
                {"type": "text", "text": "这张图片中有什么？"},
                {"type": "image_url", "image": base64_image}
            ]
        )
    ]
)
print(response.content)
```

输入的格式化

```python
import base64
import os
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage   # 补上导入

load_dotenv(override=True)
CLOSEAI_API_KEY = os.getenv("CLOSEAI_API_KEY")
CLOSEAI_BASE_URL = os.getenv("CLOSEAI_BASE_URL")

model = init_chat_model(
    model="claude-haiku-4-5",
    model_provider="openai",
    temperature=0.7,          # ← 关键修正：0~1 之间
    api_key=CLOSEAI_API_KEY,
    base_url=CLOSEAI_BASE_URL,
)


def encode_image(img_path):
    with open(img_path, "rb") as img_file:
        return base64.b64encode(img_file.read()).decode('utf-8')


img_path = "avatar.jpg"
base64_image = encode_image(img_path)

response = model.invoke([
    HumanMessage(content=[
        {"type": "text", "text": "这张图片中有什么？"},
        {
            "type": "image",
            "base64": base64_image,
            "mime_type": "image/jpeg",
        },
    ])
])

print(response.content)
```

输出格式化

```python
import os

from dotenv import load_dotenv
from langchain.chat_models.base import init_chat_model

# 读取.env配置文件中的信息，相关环境变量以.env文件中优先
load_dotenv(override=True)
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 模型初始化
llm_deepseek = init_chat_model(
    model="deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    api_base=DEEPSEEK_BASE_URL,
    extra_body={"thinking": {"type": "enabled"}}
)

# 模型调用
response = llm_deepseek.invoke("请用一句话介绍自己")
print("="*20, " -> response ", "="*20)
print(response)
print("="*20, " -> response.content ", "="*20)
print(response.content)
print("="*20, " -> response.content_blocks ", "="*20)
print(response.content_blocks)
```
