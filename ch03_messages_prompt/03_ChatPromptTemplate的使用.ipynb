结合大模型

````python
import os

from dotenv import load_dotenv
from langchain.chat_models.base import init_chat_model
from langchain_core.prompts import ChatPromptTemplate

# 读取.env配置文件中的信息，相关环境变量以.env文件中优先
load_dotenv(override=True)
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

# 模型初始化
llm_deepseek = init_chat_model(
    model="deepseek-flash",
    api_key=DEEPSEEK_API_KEY,
    api_base=DEEPSEEK_BASE_URL,
    extra_body={"thinking": {"type": "enabled"}}
)

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system", "你是一个凶猛的AI助手，你的名字叫{name}"),
    ("human", "你好，最近怎么样子？"),
    ("ai", "好你妈个头"),
    ("human", "{user_input}")
])

# 提示词调用
prompt_value = chat_prompt_template.invoke({"name": "无助龙", "user_input": "聊聊你的公螳螂理论"})

# 模型调用
response = llm_deepseek.invoke(prompt_value)
print(response)
````

可复用的模板文件

````python
from langchain_core.prompts import ChatPromptTemplate

class PromptLibrary:
    """可复用的提示词模板库"""

    TRANSLATOR = ChatPromptTemplate.from_messages([
        ("system", "你是专业翻译，精通{source_lang}和{target_lang}"),
        ("user", "翻译以下文本：\n{text}")
    ])

    CODE_REVIEWER = ChatPromptTemplate.from_messages([
        ("system", "你是{language}代码审查专家，重点关注{focus}"),
        ("user", "审查代码：\n```{language}\n{code}\n```")
    ])

    SUMMARIZER = ChatPromptTemplate.from_messages([
        ("system", "你是内容摘要专家"),
        ("user", "将一下内容总结为{num}个要点：\n{content}")
    ])

    TUTOR = ChatPromptTemplate.from_messages([
        ("system", "你是{subject}导师，学生水平：{level}"),
        ("user", "{question}")
    ])
````
